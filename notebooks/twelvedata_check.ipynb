```python
import os
import requests
from dotenv import load_dotenv

load_dotenv()

API_KEY = os.getenv("TWELVEDATA_API_KEY")
base = "https://api.twelvedata.com"
```

### All symbols

```python
r = requests.get(f"{base}/forex_pairs")
data = r.json()

symbols = [item["symbol"] for item in data["data"]]
symbols
```

### Api usage

```python
r = requests.get(f"{base}/api_usage", params={"apikey": API_KEY})
r.json()
```

### Earliest timestamp

```python
r = requests.get(f"{base}/earliest_timestamp", params={
    "symbol": "EUR/USD",
    "interval": "1day",
    "timezone": "UTC",
    "apikey": API_KEY
})

r.json()
```

### Historical data

```python
r = requests.get(f"{base}/time_series", params={
    "symbol": "EUR/USD",
    "interval": "1h",
    "start_date": "2025-01-01",
    "end_date": "2025-01-10",
    "timezone": "UTC",
    "apikey": API_KEY
})

data = r.json()
data
```

### Realtime quotes


```python
r = requests.get(f"{base}/quote", params={
    "symbol": "EUR/USD",
    "interval": "1min",
    "timezone": "UTC",
    "apikey": API_KEY
})

r.json()
```
